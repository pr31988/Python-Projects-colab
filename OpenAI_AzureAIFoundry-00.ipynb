```python
import os
from openai import AzureOpenAI
from google.colab import userdata

endpoint = userdata.get('AZURE-ENDPOINT-URL')
model_name = "gpt-5-mini"
deployment_name = "gpt-5-mini"
subscription_key = userdata.get('AZURE-OPENAI-API-KEY')
api_version = '2024-12-01-preview'

client = AzureOpenAI(
    api_version=api_version,
    azure_endpoint=endpoint,
    api_key=subscription_key,
)

response = client.chat.completions.create(
            messages=[{"role": "system",
                       "content": "You are a Travel planner."},
                      {"role": "user",
                       "content": "I am going to Japan, what should I see in 2 weeks? add time for theme park"}],
            max_completion_tokens=4128,
            model=deployment_name)

print(response.choices[0].message.content)
```
